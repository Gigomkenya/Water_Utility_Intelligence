
# Water Utility Performance and NRW Intelligence

## Customer Data Inspection and Quality Report

This notebook examines the cleaned customer dataset used in the
Water Utility Performance and NRW Intelligence project.

**Dataset:** customers_final.csv  
**Expected records:** 5,000  
**Source:** Mockaroo-generated simulated customer data.

The notebook checks dataset structure, missing values, duplicate
records, customer ID uniqueness and customer distributions.



## 1. Import Required Libraries


```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
```


## 2. Load the Final Cleaned Dataset


```python

file_path = Path("data") / "customers_final.csv"

if not file_path.exists():
    raise FileNotFoundError(
        f"Dataset not found: {file_path}. "
        "Check that the notebook is in the project root."
    )

df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
print("Total customer records:", len(df))
```


## 3. Inspect Dataset Structure


```python

print("Dataset dimensions:", df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

display(df.head())
```


## 4. Check Missing Values


```python

missing_values = df.isnull().sum()

missing_report = pd.DataFrame({
    "Missing Values": missing_values,
    "Missing Percentage": (missing_values / len(df) * 100).round(2)
})

display(missing_report)

print("Total missing values:", df.isnull().sum().sum())
```


## 5. Check Duplicate Records


```python

complete_duplicates = df.duplicated().sum()

print("Complete duplicate records:", complete_duplicates)
```


## 6. Validate Customer IDs


```python

id_candidates = ["customer_id", "customerid", "customer id"]

id_column = next(
    (col for col in df.columns
     if col.strip().lower() in id_candidates),
    None
)

if id_column is not None:
    print("Customer ID column:", id_column)
    print("Unique customer IDs:", df[id_column].nunique())
    print("Duplicate customer IDs:", df[id_column].duplicated().sum())
    print("Missing customer IDs:", df[id_column].isnull().sum())
else:
    print("No recognised customer ID column was found.")
```


## 7. Analyse Customer Categories


```python

category_columns = [
    "gender",
    "customer_category",
    "county",
    "service_area"
]

for column in category_columns:
    if column in df.columns:
        print(f"\nDistribution of {column}:")
        display(df[column].value_counts(dropna=False).to_frame("Count"))
```


## 8. Visualise Customer Distributions


```python

for column in ["gender", "customer_category"]:
    if column in df.columns:
        counts = df[column].value_counts()

        counts.plot(kind="bar", figsize=(8, 4))
        plt.title(f"Customer Distribution by {column.replace('_', ' ').title()}")
        plt.xlabel(column.replace("_", " ").title())
        plt.ylabel("Number of Customers")
        plt.xticks(rotation=45)
        plt.tight_layout()
        plt.show()
```


## 9. Final Data Quality Summary


```python

print("WATER UTILITY CUSTOMER DATA QUALITY REPORT")
print("=" * 45)

print("Dataset:", file_path)
print("Total records:", len(df))
print("Total columns:", len(df.columns))
print("Complete duplicate records:", df.duplicated().sum())
print("Total missing values:", df.isnull().sum().sum())

if id_column is not None:
    print("Unique customer IDs:", df[id_column].nunique())
    print("Duplicate customer IDs:", df[id_column].duplicated().sum())

print("\nData inspection completed.")
```
